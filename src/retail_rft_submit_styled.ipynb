    <h3 style="margin: 0; color: #FFFFFF; font-weight: 600;">Submit the RFT Job (warm-started from the SFT adapter above)</h3>
</div>

<p>Uploads the train data and recipe code as Foundry datasets, assembles a <code>CommandJob</code>, and submits it via the <code>azure-ai-projects</code> SDK. </p>
<p><b>Starting point &mdash; choose one:</b></p>

<p>Pass <code>cluster_name=&quot;...&quot;</code> to target a specific compute cluster you have access to. This argument is <b>required</b> &mdash; submission fails fast if it is missing, empty, or a placeholder, and the Azure SDK surfaces an error if the cluster does not exist.</p>
<p><b>Tune the run &mdash; three optional knob groups:</b> all keys are optional; omit any group to keep the recipe defaults.</p>
<ul>
<li>🎲 <b>sampling</b> &mdash; per-rollout generation. Keys: <code>rollout_temperature</code>, <code>rollout_max_response_len</code>, <code>n_samples_per_prompt</code>, <code>rollout_batch_size</code>.</li>
<li>🧠 <b>training</b> &mdash; GRPO optimizer + schedule. Keys: <code>learning_rate</code>, <code>global_batch_size</code>, <code>num_rollouts</code>, <code>kl_loss_coef</code>, <code>entropy_coef</code>.</li>
<li>⚙️ <b>rollout</b> &mdash; engine / compute layout. Keys: <code>rollout_num_gpus</code>, <code>rollout_num_gpus_per_engine</code>, <code>sglang_mem_fraction</code>, <code>max_tokens_per_gpu</code>, <code>tensor_parallel</code>.</li>
</ul>

```python
from slime_rl_setup import submit_job


# ── Sampling (how the model generates each rollout) ───────────
sampling = {
    "rollout_temperature":       "0.7",   # softmax temperature when sampling — raise for more diverse / exploratory answers, lower for sharper, more deterministic ones
    "rollout_max_response_len":  1536,    # max tokens the model may emit per response — raise to allow longer tool-call traces, lower to cap cost
    "n_samples_per_prompt":      16,      # GRPO group size — number of completions drawn per prompt; bigger groups give a lower-variance reward signal but cost more rollout compute
    "rollout_batch_size":        4,       # distinct prompts processed per rollout step — how many different scenarios the model sees before each policy update
}

# ── Training (GRPO optimizer + schedule) ──────────────────────
training = {
    "learning_rate":      "6e-7",          # peak actor LR — raise if reward improves too slowly, lower if training becomes unstable / reward collapses
    "global_batch_size":  32,              # samples aggregated per optimizer step — larger = smoother gradients and more GPU memory pressure
    "num_rollouts":       200,             # total rollout iterations the job will run — sets the overall length (and cost) of the training run
    "kl_loss_coef":       "0.03",          # strength of KL penalty toward the reference/SFT model — raise to keep the policy close to the seed (stable), lower to let it drift further
    "entropy_coef":       "0.0",           # entropy bonus weight — raise to encourage exploration, leave at 0 for pure reward maximization
}

# ── Rollout engine / compute layout (advanced) ────────────────
rollout = {
    "rollout_num_gpus":            16,    # total GPUs dedicated to the SGLang generation pool (separate from the actor GPUs)
    "rollout_num_gpus_per_engine": 8,     # tensor-parallel size of each SGLang engine — number of parallel engines = rollout_num_gpus / this
    "sglang_mem_fraction":         0.85,  # fraction of each GPU's HBM reserved for the SGLang KV cache — raise for longer contexts, lower to leave headroom
    "max_tokens_per_gpu":          1024,  # actor-side micro-batch in tokens per GPU — raise for higher throughput, lower if you hit OOM
    "tensor_parallel":             8,     # tensor-parallel size of the actor model (Qwen3-14B fits TP=8 on one 8-GPU node)
}

# warm_start=True  -> seed RFT from the SFT LoRA checkpoint URI you pass (recommended)
# warm_start=False -> cold-start RFT directly on the base model (no SFT seed; sft_lora_uri is ignored)
# verbose_logs=False (default) -> only emit eval/reward_mean, train/response_length_mean,
#                                 train/entropy_loss, train/kl_loss. Set True for the full debug metric set.
job_id = submit_job(
    cluster_name="testfoundrywcusclustera100",
    warm_start=True,
    sft_lora_uri=sft_lora_uri,   # set above by wait_for_sft_lora()
    sampling=sampling,
    training=training,
    rollout=rollout,
    verbose_logs=False,
)
```
