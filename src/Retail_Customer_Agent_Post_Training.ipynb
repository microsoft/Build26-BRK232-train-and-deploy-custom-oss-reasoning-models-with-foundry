```python
# Step 1: Kick off the SFT LoRA job and wait for the adapter URI.
from slime_rl_setup import submit_sft_lora_from_local, wait_for_sft_lora

# Data = SFT train/validation traces used to teach instruction following + format.
data = {
    "train_path":   "post-training-sft-recipe/demo-artifacts/data/retail_train_sft.jsonl",
    "val_path":     "post-training-sft-recipe/demo-artifacts/data/retail_val_sft.jsonl",
    "dataset_name": "retail-sft-qwen3-14b-a100-data",                                               # Foundry dataset name; a new version is created per run
}

# Code = training script package uploaded as a code asset.
code = {
    "code_dir":          "post-training-sft-recipe/demo-artifacts/code",
    "code_dataset_name": "retail-sft-qwen3-14b-a100-code",
}

# Compute = where and how the SFT fine-tuning job runs.
compute = {
    "cluster_name":  "Atlas100",                                                                    # A100 compute registered in this Foundry project
    "instance_type": "Singularity.ND96am_A100_v4-n1",                                               # single-node A100 SKU
    "image":         "mcr.microsoft.com/azureml/curated/acft-hf-nlp-gpu:119",
}

# Model = base checkpoint + SFT run length for LoRA warm-start.
model = {
    "base_model": "Qwen/Qwen3-14B",                                                                 # any HF model id your image can load
    "epochs":     3,                                                                                # LoRA warm-start length
}

sft_job_id = submit_sft_lora_from_local(
    # Data
    train_path=data["train_path"],
    val_path=data["val_path"],
    dataset_name=data["dataset_name"],
    # Code
    code_dir=code["code_dir"],
    code_dataset_name=code["code_dataset_name"],
    # Compute
    cluster_name=compute["cluster_name"],
    instance_type=compute["instance_type"],
    image=compute["image"],
    # Model
    base_model=model["base_model"],
    epochs=model["epochs"],
)

# Wait until the SFT job reaches a terminal state, then use LoRA URI for RFT warm-start.
# `azureai://.../data/<checkpoints>/versions/<v>` URI for the LoRA adapter.
sft_lora_uri = wait_for_sft_lora(sft_job_id)
print("SFT_LORA_URI =", sft_lora_uri)
```

```python
# Step 2: Launch one post-training job on your cluster.
from slime_rl_setup import submit_job


# Sampling = how the policy explores candidate answers before grading.
sampling = {
    "rollout_temperature":       "0.7",   # randomness · raise -> broader exploration · too far -> noisy tool choices
    "rollout_max_response_len":  1536,    # response token cap · raise -> longer traces · too far -> slower, costlier
    "n_samples_per_prompt":      4,       # candidates per prompt · raise -> stronger ranking signal · too far -> GPU cost spike
    "rollout_batch_size":        4,       # prompts per step · raise -> more coverage · too far -> memory pressure
}

# Training = how aggressively the policy updates from scored rollouts.
training = {
    "learning_rate":      "6e-7",         # step size · raise -> faster movement · too far -> oscillation, reward collapse
    "global_batch_size":  32,             # samples per optimizer step · raise -> smoother gradients · too far -> throughput limits
    "num_rollouts":       100,            # total training iterations · raise -> more improvement · too far -> diminishing returns
    "kl_loss_coef":       "0.03",         # drift penalty · raise -> safer updates · too far -> stuck behavior
    "entropy_coef":       "0.0",          # exploration bonus · raise -> more exploration · too far -> weaker convergence
}

# Rollout compute = how GPU engines generate trajectories for grading.
rollout = {
    "rollout_num_gpus":            16,    # generation pool size · raise -> more parallel samples · too far -> cluster contention
    "rollout_num_gpus_per_engine": 8,     # TP width per engine · raise -> bigger shards · too far -> underutilized
    "sglang_mem_fraction":         0.85,  # KV cache fraction · raise -> longer contexts · too far -> OOM
    "max_tokens_per_gpu":          1024,  # actor tokens per GPU step · raise -> higher throughput · too far -> OOM
    "tensor_parallel":             8,     # actor shard count · raise -> larger weights fit · too far -> comm overhead
}

# warm_start=True  -> seed from the SFT LoRA at sft_lora_uri (faster reward gains)
# warm_start=False -> cold-start from the base model (sft_lora_uri ignored)
job_id = submit_job(
    cluster_name="Hyperion100",                                     # H100 SKU registered in this Foundry project
    instance_type="Singularity.ND96r_H100_v5",                      # single-node H100 SKU
    warm_start=False,
    sft_lora_uri="<azureai://accounts/.../data/.../versions/...>",  # used only when warm_start=True
    sampling=sampling,
    training=training,
    rollout=rollout,
    verbose_logs=False,
)
# Expect: provisioning logs, then repeating reward/KL trends, then a terminal status.
```
